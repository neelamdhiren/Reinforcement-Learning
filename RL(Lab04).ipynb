```python
import math

machines = ['A', 'B', 'C']

N_play = 10

N = [0, 0, 0]
R = [0, 0, 0]
UCB = [0, 0, 0]
Q = [0, 0, 0]

total_reward = 0

print("===== UCB Multi-Armed Bandit =====")

for play in range(1, N_play + 1):

    if play <= 3:
        selected = play - 1
    else:
        for i in range(len(machines)):
            UCB[i] = Q[i] + math.sqrt(
                (2 * math.log(play)) / N[i]
            )

        selected = UCB.index(max(UCB))

    print("\nPlay", play)
    print("Selected Machine:", machines[selected])

    reward = float(input("Enter reward: "))

    N[selected] += 1
    R[selected] += reward

    Q[selected] = Q[selected] + (
        (reward - Q[selected]) / N[selected]
    )

    total_reward += reward

    print("Updated Q-value:", round(Q[selected], 3))

for i in range(len(machines)):
    if N[i] > 0:
        UCB[i] = Q[i] + math.sqrt(
            (2 * math.log(N_play)) / N[i]
        )
    else:
        UCB[i] = float('inf')

print("\n====================================")
print("          FINAL RESULTS")
print("====================================")

print("\nMachine\tN\tR\tUCB\tQ")

for i in range(len(machines)):
    print(
        machines[i],
        "\t", N[i],
        "\t", round(R[i], 2),
        "\t", round(UCB[i], 3),
        "\t", round(Q[i], 3)
    )

print("\nTotal Reward Received:", round(total_reward, 2))
```
