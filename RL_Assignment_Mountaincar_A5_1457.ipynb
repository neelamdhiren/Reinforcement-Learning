```python
import gymnasium as gym
env = gym.make("MountainCar-v0")
print("Environment created")
```

```python
state, info = env.reset()
print("Car position:", state[0])
print("Car velocity:", state[1])
```

```python
action = env.action_space
print(action)
```

```python
action = env.action_space.sample()
if action == 0:
    print("Accelerate left direction")
if action == 1:
    print("No Accelerate")
if action == 2:
    print("Accelerate right direction")
next_state, reward, terminated, truncated, info = env.step(action)
print("Current state:", state)
print("Next state:", next_state)
print("Reward:", reward)
print("Terminated:", terminated)
print("Truncated:", truncated)
```

```python
current_state, info = env.reset()
print("-" * 40)
print("Current State Information: Reset Environment")
print("-" * 40)
print(f"Car Position: {current_state[0]}")
print(f"Car Velocity: {current_state[1]}")
print("-" * 40)
episodes = 20
total_reward = 0
for episode in range(episodes):
    action = env.action_space.sample()
    if action == 0:
        print("Accelerate left direction")
    if action == 1:
        print("No Accelerate")
    if action == 2:
        print("Accelerate right direction")
    next_state, reward, terminated, truncated, info = env.step(action)
    print(
        f"episode={episode} | next_state={next_state} | "
        f"current_state={current_state} | reward={reward} | "
        f"terminated={terminated}"
    )

    total_reward += reward

    if terminated or truncated:
        current_state, info = env.reset()
    else:
        current_state = next_state

print(f"Total reward={total_reward}")
print(f"Average reward={total_reward / (episodes + 1)}")
```
