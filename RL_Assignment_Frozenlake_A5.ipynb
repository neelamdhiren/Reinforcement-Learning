```python
import gymnasium as gym
env = gym.make("FrozenLake-v1")
print("Frozen Lake Environment is Created")
```

Reset the environment


```python
state, info = env.reset()
print("State:", state)
```

```python
action = env.action_space
print(action)
```

**Agent act**

```python
action = env.action_space.sample()
if action == 0:
    print("Move left")
if action == 1:
    print("Move down")
if action == 2:
    print("Move right")
if action == 3:
    print("Move up")
current_state = state
next_state, reward, terminated, truncated, info = env.step(action)
print("Current state:", current_state)
print("Next state:", next_state)
print("Reward:", reward)
print("Terminated:", terminated)
print("Truncated:", truncated)
```

Random agent(interacts for many steps/episodes)

```python
import gymnasium as gym
env = gym.make("FrozenLake-v1")
current_state, info = env.reset()
print("Current state:", current_state)
total_reward = 0
episodes = 20
for episode in range(episodes):
    action = env.action_space.sample()
    next_state, reward, terminated, truncated, info = env.step(action)
    print(
        f"episode={episode} | next_state={next_state} | "
        f"current_state={current_state} | reward={reward} | "
        f"terminated={terminated} | truncated={truncated} | "
        f"action={action}"
    )
    total_reward += reward
    if terminated or truncated:
        break
    current_state = next_state
average_reward = total_reward / (episodes + 1)
print(f"Total reward={total_reward} | Average reward={average_reward}")
```
