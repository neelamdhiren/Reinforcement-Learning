Action implementation

```python
import gymnasium as gym

# Create CartPole environment
env = gym.make("CartPole-v1")
print("Cart Pole environment is created")

# Reset environment
current_state, info = env.reset()

print("=" * 50)
print("Initial State Information")
print("=" * 50)
print("Current State:", current_state)
print("Cart Position:", current_state[0])
print("Cart Velocity:", current_state[1])
print("Pole Angle:", current_state[2])
print("Pole Velocity:", current_state[3])
print("=" * 50)

# Get action space
action = env.action_space
print(action)
action = env.action_space.sample()
if action == 0:
    print("Cart moves left")
if action == 1:
    print("Cart moves right")
next_state, reward, terminated, truncated, info = env.step(action)
print("=" * 50)
print("Next State Information")
print("=" * 50)
print("Next State:", next_state)
print("Next Cart Position:", next_state[0])
print("Next Cart Velocity:", next_state[1])
print("Next Pole Angle:", next_state[2])
print("Next Pole Velocity:", next_state[3])
print("Reward:", reward)
print("Terminated:", terminated)
print("Truncated:", truncated)
print("=" * 50)
env.close()
```

Random agent(interacts for many steps/episodes)


```python
import gymnasium as gym
env = gym.make("CartPole-v1")
print("CartPole Environment Created")
episodes = 20
total_reward = 0
for episode in range(episodes):
    current_state, info = env.reset()
    episode_reward = 0
    steps = 0
    print("\n" + "=" * 50)
    print("Episode:", episode + 1)
    print("=" * 50)
    while True:
        # Select a random action
        action = env.action_space.sample()
        if action == 0:
            print("Cart moves left")
        elif action == 1:
            print("Cart moves right")
        next_state, reward, terminated, truncated, info = env.step(action)
        print("Current State:", current_state)
        print("Next State:", next_state)
        print("Reward:", reward)
        episode_reward += reward
        steps += 1
        current_state = next_state
        if terminated or truncated:
            break
    total_reward += episode_reward
    print("Episode Steps:", steps)
    print("Episode Total Reward:", episode_reward)
    print("Episode Finished")
average_reward = total_reward / episodes
print("\n" + "=" * 50)
print("FINAL RESULTS")
print("=" * 50)
print("Number of Episodes:", episodes)
print("Total Reward:", total_reward)
print("Average Reward:", average_reward)
env.close()
```
